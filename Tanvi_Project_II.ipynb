```python
!pip install pandas
!pip install numpy
!pip install matplotlib
!pip install seaborn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
uploaded = files.upload()

df = pd.read_csv("Retail_sales_data.csv")
df.head()
#step1:


print("\n--- Dataset Head ---")
print(df.head())

print("\n--- Dataset Info ---")
print(df.info())

print("\n--- Dataset Description ---")
print(df.describe())

#step2:
df = pd.read_csv(file_path)

print("\n--- Dataset Head ---")
print(df.head())

print("\n--- Dataset Info ---")
print(df.info())

print("\n--- Dataset Description ---")
print(df.describe())

#step3:
payment_trends = df['Payment Method'].value_counts()
print("\n--- Most Common Payment Methods ---")
print(payment_trends)

# 2️⃣ Top Cities by Order Count
city_orders = df['City'].value_counts()
print("\n--- Orders by City ---")
print(city_orders)

#step4:
high_price_low_sales = df.groupby('Product Name').agg({
    'Unit Price':'mean',
    'Quantity Sold':'sum'
}).reset_index()

# Define threshold
price_threshold = high_price_low_sales['Unit Price'].quantile(0.75)
low_sales_threshold = high_price_low_sales['Quantity Sold'].quantile(0.25)

premium_low = high_price_low_sales[
    (high_price_low_sales['Unit Price'] >= price_threshold) &
    (high_price_low_sales['Quantity Sold'] <= low_sales_threshold)
]

print("\n--- Premium but Low Sales Products ---")
print(premium_low.sort_values('Unit Price', ascending=False))
category_segment = df.groupby(['Customer Segment','Product Category'])['Quantity Sold'].sum().reset_index()
print("\n--- Category Sales by Customer Segment ---")
print(category_segment)

#step 5:
top_spenders = df.groupby('Customer Name')['Total Sales Amount'].sum().sort_values(ascending=False)
print("\n--- Top Spending Customers ---")
print(top_spenders.head(10))

# ➤ Repeat vs One Time Buyer
customer_counts = df.groupby('Customer Name')['Order ID'].nunique().reset_index()
customer_counts.columns = ['Customer Name','Order Count']

repeat_buyers = customer_counts[customer_counts['Order Count']>1]
one_time_buyers = customer_counts[customer_counts['Order Count']==1]

print(f"\nRepeat Buyers: {len(repeat_buyers)}")
print(f"One Time Buyers: {len(one_time_buyers)}")

#step6:
plt.style.use('seaborn')

# 🟦 Line Chart: Daily Order Volume
daily_orders = df.set_index('Date').resample('D')['Order ID'].count().reset_index()
plt.figure(figsize=(12,6))
sns.lineplot(x='Date', y='Order ID', data=daily_orders)
plt.title("📈 Daily Order Volume")
plt.xlabel("Date")
plt.ylabel("Number of Orders")
plt.show()

# 🟩 Bar Plot: Top 10 Products by Total Revenue
top_products_rev = df.groupby('Product Name')['Total Sales Amount'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(12,6))
top_products_rev.plot(kind='bar')
plt.title("🛍️ Top 10 Products by Revenue")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()

# 🟨 Count Plot: Orders by Customer Segment
plt.figure(figsize=(8,5))
sns.countplot(x='Customer Segment', data=df)
plt.title("👥 Orders by Customer Segment")
plt.show()

#step7:
# ➤ Top 10 products by revenue
top_10_products = df.groupby('Product Name')['Total Sales Amount'].sum().sort_values(ascending=False).head(10)
top_10_products.to_csv("top_10_products_by_revenue.csv")
print("\n✔ Exported: top_10_products_by_revenue.csv")

# ➤ Country-wise AOV (Average Order Value)
country_aov = df.groupby('Country').agg({
    'Total Sales Amount':'sum',
    'Order ID':'count'
}).reset_index()
country_aov['Avg Order Value'] = country_aov['Total Sales Amount'] / country_aov['Order ID']
country_aov.to_csv("country_wise_AOV.csv")
print("✔ Exported: country_wise_AOV.csv")


print("\n🎉 ANALYSIS COMPLETE!")
```
